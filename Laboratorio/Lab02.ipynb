```python
## Laboratorio 2
# 01/09/2026
# Lucia
# Problema: Realizar un sistema que identifique si una planta necesita riego para activar el sensor
# Nivel humedad suelo
# Si el nivel de humedad es menor al 40% la planta necesita riego
# Si el nivel de humedad es igual o mayor al 40%, la planta no necesita riego
```

```python
print ("Mi sistema de riego")
humedad = float (input ("Escribe el nivel de humedad del suelo (%): "))
```

```python
print ("Dato almacenado")
print ("Humedad: ", humedad)
```

```python
# Calcular nivel de humedad
if humedad <40:
    print ("La planta necesita riego")
else:
    print ("La planta no necesita riego")
```
