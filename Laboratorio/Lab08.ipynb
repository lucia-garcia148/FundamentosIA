```python
# Importar las librerias
#Libreria utilizada para graficar
import matplotlib.pyplot as plt
#Libreria que me permite trabajar con estructuras complejas como grafos
import networkx as nx

grafo = nx.Graph()

# Creamos el grafo utilizando el metodo add_edges
grafo.add_edges_from(
    [
        ("SwitchA","PC1"),
        ("SwitchA","ImpresoraA"),
        ("SwitchA","RouterA"),
        ("RouterA","Servidor_1"),
        ("RouterA","SwitchB"),
        ("SwitchB","PC2"),
        ("SwitchB","ImpresoraA")
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size=1000,
    node_color="#e44b8d"

)
plt.show()
```
