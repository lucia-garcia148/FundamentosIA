```python
# Laboratorio 5: Grafos
# Descripción: Encuentra la ruta
# Situación: Un agente inteligente se encuentra en A y necesita llegar a G.
```

```python
# importar las librerias

 #librerias utilizadas para graficar
import matplotlib.pyplot as plt
import networkx as nx #libreria que permite trabajar con librerias complejas, como grafos.
```

```python
#creamos una instancia de la clase Graph
#Graph es una coleccion de nodos
grafo = nx.Graph()
```

```python
#creamos el grafo utilizando el método add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("A","E"),
        ("B","D"),
        ("D","F"),
        ("E","G"),
        ("E","F")
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size=1000,
    node_color="#e44b8d"

)
plt.show()
```

```python
#mostrar las rutas posibles del grafo
# Source = de donde inicia
# Target = El objetivo
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="G")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas: ")
for i,ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostar el total de pasos de cada ruta 
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
```

```python
#Aplicamos la lógica propsicional
P = pasos1 < pasos2 #P: La mejor ruta es la 1
Q = pasos2 < pasos1 #Q: La mejor ruta es la 2

if P==True: 
    print("P = True")
    print("La ruta 1 es mejor opción.")
elif Q==True:
    print("Q = True")
    print("La ruta 2 es la mejor opción")
else:
    print("P = False")
    print("Q = False")
    print("Las dos rutas son igual de buenas, haz lo que quieras!!!")
```
