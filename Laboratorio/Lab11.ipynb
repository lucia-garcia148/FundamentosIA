```python
#Grafos dirigidos
# Importar las librerias
import matplotlib.pyplot as plt #Libreria utilizada para graficar
import networkx as nx #Libreria que me permite trabajar con estructuras complejas como grafos
```

```python
#Grafo 4
grafo = nx.DiGraph()
grafo.add_weighted_edges_from([
    ("A","B",8),
    ("A","E",4),
    ("A","D",5),
    ("B","C",3),
    ("B","F",4),
    ("C","F",9),
    ("C","G",2),
    ("F","E",3),
    ("F","K",8),
    ("E","I",5),
    ("D","E",3),
    ("D","H",6),
    ("G","L",2),
    ("G","K",3),
    ("H","M",2),
    ("H","I",4),
    ("I","M",1),
    ("I","N",3),
    ("I","J",4),
    ("J","O",9),
    ("K","J",6),
    ("L","K",5),
    ("L","P",3),
    ("M","N",1),
    ("N","O",2),
    ("O","P",12),
    ("P","K",1)  
    
])
```

```python
#Imprimir el grafo
pos = nx.spring_layout( grafo, seed=12)
nx.draw(
    grafo,
    pos,
    with_labels = True,
    node_size = 1500,
    arrows = True
)
labels = nx.get_edge_attributes(grafo,"weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels = labels
)
plt.show()
```

```python
# Mostramos las rutas
rutas = list (
    nx.all_simple_paths(
        grafo,
        source = "A",
        target = "P"
    )
)
#Imprimimos las rutas
for i, ruta in enumerate (rutas, start=1):
    print (f"Ruta{i}:{ruta}")
```

```python
# Utilizamos algoritmo para obtener la ruta mas corta
ruta_mas_corta = nx.dijkstra_path(grafo, source="A", target="P", weight="weight")
tiempo_total = nx.dijkstra_path_length(grafo, source="A", target="P", weight="weight")

# Imprimir
print(f"La ruta mas corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```
