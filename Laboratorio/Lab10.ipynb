```python
#Grafos dirigidos
# Importar las librerias
import matplotlib.pyplot as plt #Libreria utilizada para graficar
import networkx as nx #Libreria que me permite trabajar con estructuras complejas como grafos
```

```python
#Creamos el grafo
grafo = nx.DiGraph()
grafo.add_weighted_edges_from([
    ("Centro", "Este", 25),
    ("Este", "Aeropuerto", 20),
    ("Centro", "Norte", 10),
    ("Norte", "Sur", 30),
    ("Centro", "Sur", 15),
    ("Sur", "Oeste", 9),
    ("Oeste", "Aeropuerto", 18),
    ("Norte", "Este", 8),
    ("Norte", "Oeste", 15)
])
```

```python
#Imprimir el grafo
pos = nx.spring_layout( grafo, seed=10)
nx.draw(
    grafo,
    pos,
    with_labels = True,
    node_size = 1500,
    arrows = True
)
labels = nx.get_edge_attributes(grafo,"weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels = labels
)
plt.show()
```

```python
# Mostramos las rutas
rutas = list (
    nx.all_simple_paths(
        grafo,
        source = "Centro",
        target = "Aeropuerto"
    )
)
# Imprimimos las rutas
for i, ruta in enumerate(rutas,start = 1):
    print(f"Ruta {i}:{ruta}" )
```

```python
#Utilizamos algoritmo para obtener la ruta más corta
ruta_mas_corta = nx.dijkstra_path( grafo,source="Centro", target="Aeropuerto",weight="weight")
tiempo_total = nx.dijkstra_path_length(grafo, source="Centro", target="Aeropuerto", weight="weight")
#Imprimir
print(f"La ruta más corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```
