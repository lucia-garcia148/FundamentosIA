```python
# Variables
servidor = False #El servidor no esta disponible
latencia = True #Supera el limite establecido
paquetes = True #Supera el limite permitido
red = True #Es anormalmente alto
servidorRespaldo = False #No Disponible
sistema = True #No esta activo
```

```python
### Reglas 
El servidor no esta disponible Y el servidor de respaldo no esta disponible Y la latencia supera
el limite Y los paquetes supera el limite Y la red es anormalmente alta Y el sistema no esta activo
```

```python
servidorActivo = (servidor== False and latencia== True and paquetes==True and red==True and
                  servidorRespaldo== False and sistema==True)
```
